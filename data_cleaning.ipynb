```python
import pandas as pd
import numpy as np
data = (pd.read_csv('data.csv'))
data = data.dropna(subset=['total_conversion', 'approved_conversion'])
data.head()
```

```python
data['CTR'] = (data['clicks']/data['impressions'])*100
data['ROI'] = ((data['total_conversion']-data['spent'])/ data['spent']) * 100
data.head()
```

```python
data = data.dropna(subset=['ROI'])
data.isnull().sum()
```

```python
data['ROI'].replace([np.inf,- np.inf],0, inplace=True)
data.head()
```

```python
data.to_csv('cleaned_data.csv', index=False)
```
